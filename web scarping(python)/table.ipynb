```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

```python
url = "https://www.nseindia.com/"  # Replace with the website URL


headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/96.0.4664.45 Safari/537.36"
}

response = requests.get(url, headers=headers)

if response.status_code == 200:
    print("Page fetched successfully!")
else:
    print(f"Failed to fetch page: {response.status_code}")
    exit()
```

```python
soup = BeautifulSoup(response.text, "html.parser")

# Find all tables
tables = soup.find_all("table")

print(f"Found {len(tables)} tables on the webpage.")  # Check number of tables found
```

```python
for idx, table in enumerate(tables):
    rows = table.find_all("tr")
    table_data = []

    # Extract table rows
    for row in rows:
        cells = [cell.text.strip() for cell in row.find_all(["td", "th"])]
        table_data.append(cells)

    # Determine the maximum number of columns
    max_cols = max(len(row) for row in table_data)

    print(max_cols)

    # Pad rows to match the max number of columns
    for row in table_data:
        while len(row) < max_cols:
            row.append("")  # Fill missing values with empty strings

    # Define headers (first row) and data (remaining rows)
    headers = table_data[0] if table_data else []
    data = table_data[1:] if len(table_data) > 1 else []

    # Convert to Pandas DataFrame
    df = pd.DataFrame(data, columns=headers if headers else [f"Column {i+1}" for i in range(max_cols)])

    # Save to CSV
    csv_filename = f"table_{idx+1}.csv"
    df.to_csv(csv_filename, index=False, encoding="utf-8")

    print(f"Table {idx+1} saved as {csv_filename}")
```
